Kaggle-ready entry point for the Robotics website. This notebook is intentionally kept under Kaggle's 1 MB GitHub-import limit.

**Projects:** 12–18

**Theme:** Light / Dark

When this notebook is run, it downloads the complete website from the public GitHub repository and restores it in `/kaggle/working/Robotics`.

````python
# Restore the complete Robotics website from GitHub.
from pathlib import Path
import shutil
import subprocess

REPO = "https://github.com/shadman-sharif/Robotics_Moving.git"
CLONE_DIR = Path("/kaggle/working/Robotics_Moving")
SITE_DIR = Path("/kaggle/working/Robotics")

if CLONE_DIR.exists():
    shutil.rmtree(CLONE_DIR)
if SITE_DIR.exists():
    shutil.rmtree(SITE_DIR)

result = subprocess.run(
    ["git", "clone", "--depth", "1", REPO, str(CLONE_DIR)],
    text=True,
    capture_output=True,
)

if result.returncode != 0:
    raise RuntimeError(
        "GitHub download failed. In Kaggle, open Notebook options and enable Internet, "
        "then run this cell again.\n\n" + result.stderr
    )

source = CLONE_DIR / "Robotics"
if not source.exists():
    raise FileNotFoundError("The GitHub repository does not contain the expected Robotics/ website folder.")

shutil.copytree(source, SITE_DIR)
print(f"Website restored to: {SITE_DIR}")
print(f"Files restored: {sum(1 for p in SITE_DIR.rglob('*') if p.is_file())}")
````

````python
# Verify the important website files.
from pathlib import Path

required = [
    Path('/kaggle/working/Robotics/index.html'),
    Path('/kaggle/working/Robotics/css/style.css'),
    Path('/kaggle/working/Robotics/js/app.js'),
    Path('/kaggle/working/Robotics/images'),
]

missing = [str(p) for p in required if not p.exists()]
if missing:
    raise FileNotFoundError("Missing required website files:\n" + "\n".join(missing))

print("✓ Robotics website is ready in /kaggle/working/Robotics")
print("✓ index.html, CSS, JavaScript and image assets are present.")
````
